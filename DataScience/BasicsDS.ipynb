```python
# TASK2: Using Def Function make(Fibonacci series) 

def fibonacci(n):
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    elif n == 2:
        return[0,1]
    else:
        fib_series = [0,1]
        for i in range(2,n):
            next_number = fib_series[i -1] + fib_series[i -2]
            fib_series.append(next_number)
        return fib_series


num = int(input("Enter a number to find Fibonacci series and Factorial: "))
fib_series = fibonacci(num)
print("Fibonacci Series:", fib_series)  


    
    

```

```python
#Task3: FIND fACTORIAL OF A NUMBER USING RECURSION:
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)
num = int(input("Enter a number to find Factorial: "))
result = factorial(num)
print("Factorial of", num, "is", result)
```

```python
##Take input and generate table according to given number

def generate_table(n):
    table = []
    for i in range(1, 11):
        table.append(f"{n} x {i} = {n * i}")
    return table
num = int(input("Enter a number to generate its table: "))
table = generate_table(num)
print("Table of", num, "is:", table)





```

```python
# Generate table according to given number suppose generate table of 1 to 20 
# it takes two input from users that how many tables you want to generate and 
# up to which number you want to generate table
def generate_tables(num_tables, up_to):
    tables ={}
    for n in range(1,num_tables + 1):
        table = []
        for i in range(1, up_to + 1):
            table.append(f"{n} x {i} = {n * i}")
        tables[n] = table
    return tables
num_tables = int(input("Enter how many tables you want to generate: "))
up_to = int(input("Enter up to which number you want to generate table: "))
tables = generate_tables(num_tables, up_to)
for n, table in tables.items():
    print(f"Table of {n}:")
    for line in table:
        print(line)
    print()  # Print a blank line between tables

    
```

## Making objects, methods and class


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Parent class (Base class)
class demo:
    # Class variables (optional)
    x = 0
    y = 0
    
    # Constructor
    def __init__(self, m, n):
        self.x = m
        self.y = n
        print(f"Parent class initialized with x={m}, y={n}")
    
    # Method in parent class
    def add(self):
        Sum = self.x + self.y
        print(f"Sum: {self.x} + {self.y} = {Sum}")
        return Sum

# Child class inheriting from demo - DEFINED OUTSIDE parent class
class b(demo):  # b inherits from demo
    # Child class can have its own constructor
    def __init__(self, m, n, p=0):
        # Call parent class constructor
        super().__init__(m, n)
        self.p = p  # Additional attribute in child class
        print(f"Child class initialized with extra parameter p={p}")
    
    # New method in child class
    def subtract(self):
        Subtract = self.x - self.y
        print(f"Subtract: {self.x} - {self.y} = {Subtract}")
        return Subtract
    
    # Override parent method (optional)
    def product(self):
        Product = self.x * self.y
        print(f"Product: {self.x} × {self.y} = {Product}")
        return Product
    
    # Additional method using all attributes
    def calculate_all(self):
        print(f"\nCalculations for x={self.x}, y={self.y}, p={self.p}:")
        self.add()
        self.subtract()
        self.product()

# Object creation - Parent class object
print("="*50)
print("PARENT CLASS DEMO")
print("="*50)
obj_parent = demo(5, 10)
obj_parent.add()
# obj_parent.product()  # This will ERROR - product is not in parent class

# Object creation - Child class object
print("\n" + "="*50)
print("CHILD CLASS DEMO")
print("="*50)
obj_child = b(5, 10, 3)  # Child object with extra parameter

# Call methods on child object
obj_child.add()      # Inherited from parent
obj_child.subtract() # Defined in child
obj_child.product()  # Overridden in child
obj_child.calculate_all()  # Child-specific method
```

```python
# Using Imputation to handle missing values in a dataset:
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.impute import KNNImputer

#using SimpleImputer to fill missing values with the mean of the column
data = np.array([[1, 2, np.nan], [4, np.nan, 6], [7, 8, 9]])
impute= SimpleImputer(strategy='median') # we can use mean too
imputed_data = impute.fit_transform(data)
print("Imputed Data using SimpleImputer (Mean):")
print(imputed_data)

#using KNNImputer to fill missing values based on the nearest neighbors
Knn_imputer = KNNImputer(n_neighbors=2)
imputed_data_knn = Knn_imputer.fit_transform(data)
print("\nImputed Data using KNNImputer:")
print(imputed_data_knn)

# filling missing values without using imputation techniques (e.g., using mean of the column)
# use Dataframes to find missing values
import numpy as np
import pandas as pd

Ds = pd.DataFrame({
    "over": [1,7,3,9,5],
    "run": [2,8,4,10,6],

})
print("Using isnull() to find missing values:")
print(Ds.isnull())
print("using isnull().sum() to count missing values in each column:")
print(Ds.isnull().sum())
print("\nDataFrame with Missing Values:")
print(Ds)



```
